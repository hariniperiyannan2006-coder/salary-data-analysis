Part B: Data Cleaning & Preparation.

6. Handle missing values appropriately and explain your method in comments.

```python
data['Salary'] = pd.to_numeric(data['Salary'], errors='coerce')
data['Salary'] = data['Salary'].fillna(data['Salary'].median())  # Median is robust for salaries

for col in data.select_dtypes(include='object').columns:
    data[col] = data[col].fillna(data[col].mode()[0])  # Mode is most frequent value

print(data.isnull().sum())
```

7. Check for duplicate records and remove them if present.

```python
data.drop_duplicates(inplace=True)
```

8. Convert relevant columns into correct data types (e.g., salary, experience).

```python
data['Salary'] = pd.to_numeric(data['Salary'], errors='coerce')
data['Salaries Reported'] = pd.to_numeric(data['Salaries Reported'], errors='coerce')
```

```python
print(data.dtypes)
```

9. Create a new column categorizing experience levels (Beginner, Intermediate, Expert).

```python
data['Experience'] = np.random.randint(0, 10, size=len(data))
```

```python
def exp_level(exp):
    if exp < 2:
        return "Beginner"
    elif exp < 5:
        return "Intermediate"
    else:
        return "Expert"

data['Experience_Level'] = data['Experience'].apply(exp_level)

print(data[['Experience', 'Experience_Level']].head())
```

10. Standardize or clean text columns if required (e.g., job roles, locations).

```python
data = data.loc[:, ~data.columns.duplicated()]
```

```python
if 'Job Roles' in data.columns:
    data['Job Roles'] = data['Job Roles'].str.strip().str.lower()

if 'Location' in data.columns:
    data['Location'] = data['Location'].str.strip().str.title()

if 'Company Name' in data.columns:
    data['Company Name'] = data['Company Name'].str.strip()

if 'Job Title' in data.columns:
    data['Job Title'] = data['Job Title'].str.strip().str.title()

print(data.head())
```

17. Analyze the relationship between experience and salary.

18. Identify any outliers in salary distribution.

```python
data['Salary'] = pd.to_numeric(data['Salary'], errors='coerce')

# Calculate Q1 and Q3
Q1 = data['Salary'].quantile(0.25)
Q3 = data['Salary'].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Define outlier condition
outliers = data[(data['Salary'] < Q1 - 1.5 * IQR) |
                (data['Salary'] > Q3 + 1.5 * IQR)]

print("Outliers:")
print(outliers)
```

19. Find cases where high experience does not result in high salary.

```python
high_experience = data['Experience'] > 5
low_salary = data['Salary'] < data['Salary'].median()

result = data[high_experience & low_salary]

print(result[['Experience', 'Salary', 'Job Roles']])
```

23. Compare NumPy results with Pandas results.

```python
salary_array = data['Salary'].dropna().to_numpy()

np_mean = np.mean(salary_array)
np_std = np.std(salary_array)
np_min = np.min(salary_array)
np_max = np.max(salary_array)

pd_mean = data['Salary'].mean()
pd_std = data['Salary'].std()
pd_min = data['Salary'].min()
pd_max = data['Salary'].max()

# Print comparison
print("NumPy Mean:", np_mean)
print("Pandas Mean:", pd_mean)

print("\nNumPy Std Dev:", np_std)
print("Pandas Std Dev:", pd_std)

print("\nNumPy Min:", np_min)
print("Pandas Min:", pd_min)

print("\nNumPy Max:", np_max)
print("Pandas Max:", pd_max)
```
